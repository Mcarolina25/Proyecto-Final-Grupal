```python
def process_metadata(df):
    # Función para extraer opciones de servicio
    def extract_service_options(misc):
        if isinstance(misc, dict) and 'Service options' in misc:
            return ', '.join(misc['Service options'])
        return None

    # Función para extraer información de salud y seguridad
    def extract_health_safety(misc):
        if isinstance(misc, dict) and 'Health & safety' in misc:
            return ', '.join(misc['Health & safety'])
        return None

    # Función para extraer información de accesibilidad
    def extract_accessibility(misc):
        if isinstance(misc, dict) and 'Accessibility' in misc:
            return ', '.join(misc['Accessibility'])
        return None

    # Función para extraer información de planificación
    def extract_planning(misc):
        if isinstance(misc, dict) and 'Planning' in misc:
            return ', '.join(misc['Planning'])
        return None

    # Función para extraer información de pagos
    def extract_payments(misc):
        if isinstance(misc, dict) and 'Payments' in misc:
            return ', '.join(misc['Payments'])
        return None

    # Aplicar las funciones de extracción al DataFrame
    df['service_options'] = df['MISC'].apply(extract_service_options)
    df['health_safety'] = df['MISC'].apply(extract_health_safety)
    df['accessibility'] = df['MISC'].apply(extract_accessibility)
    df['planning'] = df['MISC'].apply(extract_planning)
    df['payments'] = df['MISC'].apply(extract_payments)

    # Eliminar la columna 'MISC' original
    df = df.drop('MISC', axis=1)

    return df

df = process_metadata(df)

print(df.head())
print(df.info())
```

```python
# Define la ruta al archivo user.parquet
directorio = r'c:\Users\segen\Documents\Henry\Proyecto3-Grupo\PF grupo\Yelp'
archivo_user = os.path.join(directorio, 'user.parquet')

# Carga el archivo user.parquet en un DataFrame de pandas
df_user = pd.read_parquet(archivo_user)

# Imprime información sobre el DataFrame resultante
print(f"DataFrame de usuarios creado con {len(df_user)} registros.")
print(df_user.head())
print(df_user.info())

# Muestra las columnas del DataFrame
print("\nColumnas del DataFrame:")
print(df_user.columns)

# Explora algunas columnas específicas y sus tipos de datos
print("\nEjemplo de la columna 'user_id':")
print(df_user['user_id'].head())

print("\nEjemplo de la columna 'name':")
print(df_user['name'].head())

print("\nEjemplo de la columna 'friends':")
print(df_user['friends'].head())

# Función para calcular la cantidad de amigos
def contar_amigos(lista_amigos):
    """
    Calcula la cantidad de amigos de un usuario.

    Args:
        lista_amigos (str): Una cadena que contiene la lista de amigos, separados por comas.

    Returns:
        int: El número de amigos. Retorna 0 si la lista es vacía o no es una cadena.
    """
    if isinstance(lista_amigos, str):
        amigos = lista_amigos.split(', ')
        return len(amigos)
    else:
        return 0

# Aplica la función para crear la columna 'num_amigos'
df_user['num_amigos'] = df_user['friends'].apply(contar_amigos)

# Imprime las primeras filas con la nueva columna
print("\nDataFrame con la columna 'num_amigos':")
print(df_user[['user_id', 'name', 'friends', 'num_amigos']].head())

# Función para verificar si un usuario fue miembro elite en un año específico
def fue_elite_en(lista_anios, anio):
    """
    Verifica si un usuario fue miembro elite en un año específico.

    Args:
        lista_anios (list): Una lista de años en los que el usuario fue miembro elite.
        anio (int): El año a verificar.

    Returns:
        bool: True si el usuario fue miembro elite en el año especificado, False de lo contrario.
              Retorna False si la lista no es una lista.
    """
    if isinstance(lista_anios, list):
        return anio in lista_anios
    else:
        return False

# Aplica la función para crear una columna que indica si fue elite en 2015
df_user['fue_elite_2015'] = df_user['elite'].apply(lambda x: fue_elite_en(x, 2015))

# Imprime las primeras filas con la nueva columna
print("\nDataFrame con la columna 'fue_elite_2015':")
print(df_user[['user_id', 'name', 'elite', 'fue_elite_2015']].head())

# Calcular el total de cumplidos recibidos
columnas_complidos = [
    'compliment_hot', 'compliment_more', 'compliment_profile', 'compliment_cute',
    'compliment_list', 'compliment_note', 'compliment_plain', 'compliment_cool',
    'compliment_funny', 'compliment_writer', 'compliment_photos'
]

# Suma las columnas de cumplidos para obtener el total
df_user['total_compliment'] = df_user[columnas_complidos].sum(axis=1)

# Imprime las primeras filas con la nueva columna
print("\nDataFrame con la columna 'total_compliment':")
print(df_user[['user_id', 'name'] + columnas_complidos + ['total_compliment']].head())
```

```python
def cargar_datos_json(ruta_archivo):
    """Carga datos JSON desde un archivo, manejando posibles errores de decodificación."""
    datos = []
    with open(ruta_archivo, 'r', encoding='utf-8') as archivo:
        for linea in archivo:
            try:
                registro_json = json.loads(linea)
                datos.append(registro_json)
            except json.JSONDecodeError as e:
                print(f"Error al decodificar JSON en {ruta_archivo}: {e}")
                continue
    return pd.DataFrame(datos)

def cargar_datos_parquet(ruta_archivo):
    """Carga datos Parquet desde un archivo."""
    try:
        return pd.read_parquet(ruta_archivo)
    except Exception as e:
        print(f"Error al cargar el archivo Parquet {ruta_archivo}: {e}")
        return None

def cargar_datos_pickle(ruta_archivo):
    """Carga datos Pickle desde un archivo."""
    try:
        return pd.read_pickle(ruta_archivo)
    except Exception as e:
        print(f"Error al cargar el archivo Pickle {ruta_archivo}: {e}")
        return None

def procesar_columnas_fecha(df, columna_fecha):
    """Convierte una columna de fecha a objetos datetime y extrae características."""
    df[columna_fecha] = pd.to_datetime(df[columna_fecha], errors='coerce')  # Maneja formatos de fecha inválidos
    df['año'] = df[columna_fecha].dt.year
    df['mes'] = df[columna_fecha].dt.month
    df['dia_semana'] = df[columna_fecha].dt.dayofweek  # 0 = Lunes, 6 = Domingo
    df['hora'] = df[columna_fecha].dt.hour
    return df

# Ejemplo de uso:
directorio_yelp = r'c:\Users\segen\Documents\Henry\Proyecto3-Grupo\PF grupo\Yelp'

# Carga datos de Yelp
df_business = cargar_datos_pickle(os.path.join(directorio_yelp, 'business.pkl'))
df_checkin = cargar_datos_json(os.path.join(directorio_yelp, 'checkin.json'))
df_review = cargar_datos_json(os.path.join(directorio_yelp, 'review.json'))
df_tip = cargar_datos_json(os.path.join(directorio_yelp, 'tip.json'))
df_user = cargar_datos_parquet(os.path.join(directorio_yelp, 'user.parquet'))

# Procesa las columnas de fecha
if df_checkin is not None:
    df_checkin = procesar_columnas_fecha(df_checkin, 'date')
if df_tip is not None:
    df_tip = procesar_columnas_fecha(df_tip, 'date')

# Carga datos de reviews de los estados
directorios = {
    'California': r'c:\Users\segen\Documents\Henry\Proyecto3-Grupo\PF grupo\California',
    'Carolina del Sur': r'c:\Users\segen\Documents\Henry\Proyecto3-Grupo\PF grupo\Carolina del Sur',
    'Florida': r'c:\Users\segen\Documents\Henry\Proyecto3-Grupo\PF grupo\Florida',
    'Luisiana': r'c:\Users\segen\Documents\Henry\Proyecto3-Grupo\PF grupo\Luisiana',
    'Massachusetts': r'c:\Users\segen\Documents\Henry\Proyecto3-Grupo\PF grupo\Masachusets',
    'Texas': r'c:\Users\segen\Documents\Henry\Proyecto3-Grupo\PF grupo\Texas',
    'Washington': r'c:\Users\segen\Documents\Henry\Proyecto3-Grupo\PF grupo\Washington'
}

dataframes_estados = {}
for estado, directorio in directorios.items():
    # Construir la ruta del archivo JSON para las reviews del estado
    ruta_archivo = os.path.join(directorio, 'review.json')
    
    # Intentar cargar los archivos JSON individuales si el archivo consolidado no existe
    if not os.path.exists(ruta_archivo):
        print(f"Archivo no encontrado: {ruta_archivo}. Intentando cargar archivos JSON individuales...")
        
        # Listar todos los archivos JSON en el directorio
        archivos_json = [f for f in os.listdir(directorio) if f.endswith('.json')]
        
        # Crear una lista para almacenar los DataFrames individuales
        dfs_estado = []
        
        # Iterar sobre cada archivo JSON y cargarlo en un DataFrame
        for archivo_json in archivos_json:
            ruta_archivo_individual = os.path.join(directorio, archivo_json)
            try:
                df_individual = cargar_datos_json(ruta_archivo_individual)
                dfs_estado.append(df_individual)
            except Exception as e:
                print(f"Error al cargar {archivo_json}: {e}")
        
        # Concatenar todos los DataFrames individuales en uno solo
        if dfs_estado:
            df_estado_review = pd.concat(dfs_estado, ignore_index=True)
            print(f"Se concatenaron {len(dfs_estado)} archivos JSON en un DataFrame para {estado}.")
        else:
            print(f"No se pudieron cargar archivos JSON individuales para {estado}.")
            df_estado_review = None
    else:
        # Cargar los datos JSON en un DataFrame
        df_estado_review = cargar_datos_json(ruta_archivo)
    
    # Verificar si el DataFrame se cargó correctamente
    if df_estado_review is not None:
        # Procesar las columnas de fecha
        df_estado_review = procesar_columnas_fecha(df_estado_review, 'time')
        
        # Guardar el DataFrame en el diccionario
        dataframes_estados[estado] = df_estado_review
        
        # Imprimir información sobre el DataFrame creado
        print(f"DataFrame de reviews de {estado} creado con {len(dataframes_estados[estado])} registros.")
        print(dataframes_estados[estado].head())
        print(dataframes_estados[estado].info())
    else:
        print(f"No se pudo cargar el DataFrame de reviews de {estado}.")

# Almacena todos los DataFrames en un diccionario
dataframes = {
    'business': df_business,
    'checkin': df_checkin,
    'review': df_review,
    'tip': df_tip,
    'user': df_user,
    'reviews_estados': dataframes_estados
}
```
